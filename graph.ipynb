```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.style.use("seaborn-v0_8")

sources = np.array([1,2,3])

# raw data
mp4_runs = np.array([
[5083,3696,3963],
[4002,3885,3192],
[3483,3325,3318]
])

zip_runs = np.array([
[3461,3311,3816],
[4364,3973,3497],
[3286,3186,3869]
])

# averages
mp4_avg = mp4_runs.mean(axis=1)
zip_avg = zip_runs.mean(axis=1)

# speedup
mp4_speedup = mp4_avg[0] / mp4_avg
zip_speedup = zip_avg[0] / zip_avg

# smooth x axis
x_smooth = np.linspace(1,3,200)

def smooth_curve(x,y):
    poly = np.poly1d(np.polyfit(x,y,2))
    return poly(x_smooth)
```

```python
# MP4 runs

plt.figure(figsize=(7,5))

for i in range(3):
    plt.plot(x_smooth,
             smooth_curve(sources,mp4_runs[:,i]),
             label=f"Run {i+1}",
             linewidth=2)

    plt.scatter(sources,mp4_runs[:,i])

plt.xlabel("Number of Sources")
plt.ylabel("Download Time (ms)")
plt.title("MP4 Download Time per Run")

plt.grid(True)
plt.legend()

plt.ylim(3000,5200)

plt.savefig("graph_mp4_runs.png",dpi=300)

plt.show()
```

```python
# ZIP runs

plt.figure(figsize=(7,5))

for i in range(3):
    plt.plot(x_smooth,
             smooth_curve(sources,zip_runs[:,i]),
             label=f"Run {i+1}",
             linewidth=2)

    plt.scatter(sources,zip_runs[:,i])

plt.xlabel("Number of Sources")
plt.ylabel("Download Time (ms)")
plt.title("ZIP Download Time per Run")

plt.grid(True)
plt.legend()

plt.ylim(3000,4500)

plt.savefig("graph_zip_runs.png",dpi=300)

plt.show()
```

```python
# average comparison

plt.figure(figsize=(7,5))

plt.plot(x_smooth,smooth_curve(sources,mp4_avg),label="MP4 avg",linewidth=3)
plt.plot(x_smooth,smooth_curve(sources,zip_avg),label="ZIP avg",linewidth=3)

plt.scatter(sources,mp4_avg)
plt.scatter(sources,zip_avg)

plt.xlabel("Number of Sources")
plt.ylabel("Download Time (ms)")
plt.title("Average Download Time vs Sources")

plt.grid(True)
plt.legend()

plt.ylim(3000,4500)

plt.savefig("graph_avg_curve.png",dpi=300)

plt.show()
```

```python
# speedup

plt.figure(figsize=(7,5))

plt.plot(x_smooth,
         smooth_curve(sources,mp4_speedup),
         label="MP4 speedup",
         linewidth=3)

plt.plot(x_smooth,
         smooth_curve(sources,zip_speedup),
         label="ZIP speedup",
         linewidth=3)

plt.scatter(sources,mp4_speedup)
plt.scatter(sources,zip_speedup)

plt.xlabel("Number of Sources")
plt.ylabel("Speedup")

plt.title("Speedup vs Number of Sources")

plt.grid(True)
plt.legend()

plt.savefig("graph_speedup.png",dpi=300)

plt.show()
```
